**W.1** Write a function that returns a list of the first n powers of 2 (2^0, 2^1, 2^2, 2^3...2^n).

```python
def powers_of_2(number):
    """Receives a number, n, and returns the first n powers of 2"""
    powers_results = []
    for i in range(0,number + 1):
        power = 2**i
        powers_results.append(power)
    return(powers_results)
```

**W.2** Using the function above, plot the first 100 powers of 2

```python
_100_powers_of_2 = powers_of_2(100)
print(_100_powers_of_2)
```

```python
plot = plt.plot(_100_powers_of_2)
plt.yscale('log')
plt.show()
```

**Measures of Center**

Mean and median are common measures of center in statistics. You can use them by typing:

    dataset.mean()
    dataset.median()
When analyzing skewed data, it's better to use the median as the central statistic because it is less affected by outliers.

**Measures of Spread**

Variance

    data.var()
Standard Deviation

    data.std
Quantiles/percentiles

    np.quantile(data, np.linspace(0,1, #of divisions)
Interquartile Range: helpful for boxplots and outlier detection

First define the first and third quartiles

    q1 = np.quantile(data, 0.25)
    q3 = np.quantile(data, 0.75)

Then define the IQR

    iqr = q3 - q1

Then define the thresholds

    lower = q1 - 1.5 * iqr
    lower = q3 + 1.5 * iqr

Finally, determine your outliers

    outliers = data[(data < lower) | (data > upper)]

**Probability Overview**

Proability is calculated as the number of times your outcome can occur/the total possible outcomes.

To run a random simulation of sampling, use 

    data.sample()
    data.sample( #, replace = T/F)

You can specify the number of samples as well as if the sampling will have replacement

**Discrete Distributions**

Discrete distributions are probabilities with finite outcomes. On graphs they are represented with blocks/bars, like in histograms. Uniform discrete distributions are when all outcomes have the same or very similar probabilities.

Probabilities and their statistics become more stable with increasing number of samples taken.

**Continuous Distributions**

Continuous distributions are probabilites with infinite outcomes. On graphs they are represented with lines. The area under the line will always equal 1. To deal with uniform continuous probabilities you will need to do this:
    
    from scipy.stats import uniform
For probability determination (a prob of a # and under only needs one input, but between two #s or a # and over needs two:
   
    # 5 and under
    uniform.cdf(5, min, max)
    # between 10 and 20
    uniform.cdf(20, min, max) - uniform.cdf(10, min, max)

For creating a similation:

    uniform.rvs(min, max, size = number of replicates)

**Binomial Distribution**

Binomial distributions are probabilities with exactly two outcomes. To deal with binomial probabilities you will need to do this:

    from scipy.stats import uniform

For probability determination

    For desired outcome to occur an exact number of times
    binom.pmf(#, # of trials, probability of outcome to happen once)
    For desired outcome to occur a certain number of times or under
    binom.cdf(#, # of trials, probability of outcome to happen once)

For creating a similation:

    binom.rvs(# of trials, probability, size = number of replicates)

Also good to keep in mind that probability of multiple trials can be estimated through n x p, where n is the number of trials and p is the probability of your outcome for 1 trial.
