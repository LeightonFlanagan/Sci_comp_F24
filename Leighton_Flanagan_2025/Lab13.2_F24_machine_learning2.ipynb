**How Good is Your Model?**

Types of metrics to use:
- precision: true positives/(true positives + false positives) -> how many negatives you flagged
- recall: true positives/(true positives + false negatives) -> how many positives you didn't catch
- F1 score: (precision*recall)/(precision + recall)

Putting it in practice: 

    from sklearn.metrics import confusion_matrix, classification_report

    #here we have already done the train_test_split
    knn = KNeighborsClassifier(n_neighbors = #)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)

    confusion_matrix(y_test, y_pred) #gives you the number of true positives, false positives, true negatives, and false negatives
    classification_report(y_test, y_pred) #gives you the precision, recall, and F1 scores.

    from sklearn.metrics import roc_curve
