```python
# let's go over an example plot in the old matplotlib documenation gallery

labels = ['G1', 'G2', 'G3', 'G4', 'G5']
men_means = [20, 35, 30, 35, 27]
women_means = [25, 32, 34, 20, 25]
men_std = [2, 3, 4, 1, 2]
women_std = [3, 5, 2, 3, 3]
width = 0.55       # the width of the bars: can also be len(x) sequence

fig, ax = plt.subplots() # defining a figure

ax.bar(labels, men_means, width, yerr = men_std, label = 'Men')
ax.bar(labels, women_means, width, yerr = women_std, bottom = men_means, label = "Women")

plt.ylabel('Scores')
#ax.set_ylabel('Scores')
ax.set_title('Scores by group and gender')
ax.legend()
```

```python
#stem plot
#dataset:
x = 0.5 + np.arange(8)
y = [4.8, 5.5, 3.5, 4.6, 6.5, 6.6, 2.6, 3.0]

fig,ax = plt.subplots()
ax.stem(x,y)
```

```python
#stem plot

fig,ax = plt.subplots()
ax.stem(x,y)
ax.set(ylim = (0,8))
```

```python
#stem plot

fig,ax = plt.subplots()
ax.stem(x,y, basefmt = 'C0-')
ax.set(ylim = (0,8))
```

```python
#pie chart
#dataset:
x = [1, 2, 3, 4]
colors = plt.get_cmap('Blues')(np.linspace(0.2, 0.7, len(x)))

fig, ax = plt.subplots()
ax.pie(x, colors=colors, center=(0,0))
```

```python
#pie chart

fig, ax = plt.subplots()
ax.pie(x, colors=colors, radius=1.5, center=(0,0))
```

```python
#pie chart

fig, ax = plt.subplots()
ax.pie(x, colors=colors, radius=1.5, center=(0,0))
ax.set_title("This is a Pie")
```

```python
#boxplot
#dataset: 
np.random.seed(10)
D = np.random.normal((3, 5, 4), (1.25, 1.00, 1.25), (100, 3))

fig, ax = plt.subplots()
ax.boxplot(D)
```

```python
#boxplot

fig, ax = plt.subplots()
ax.boxplot(D, patch_artist = True,
          medianprops={"color":"white"})
```

```python
#boxplot

fig, ax = plt.subplots()
ax.boxplot(D, patch_artist = True,
          medianprops={"color":"white"})
ax.axhline(y=0, color = "black")
```

```python
#quiverplot
#dataset
x = np.linspace(-4, 4, 6)
y = np.linspace(-4, 4, 6)
X, Y = np.meshgrid(x, y)
U = X + Y
V = Y - X

fig, ax = plt.subplots()
ax.quiver(X, Y, U, V)
```

```python
#quiverplot

fig, ax = plt.subplots()
ax.quiver(X, Y, U, V, color="C3")
```

```python
#quiverplot

fig, ax = plt.subplots()
ax.quiver(X, Y, U, V, color="C3")
ax.set(xlim = (-6,6), ylim = (-6,6))
```
