**E.2** What is a dictionary and why is it useful?

A dictionary is a type of variable in python that uses keys instead of indexing to call up information about that key. It's useful for containing multiple variable types rather than a numpy array, and you can call upon your specific key words to find data associated with them, which is more descriptive than indexing.

**E.3** How do you define a dictionary and access its elements? Explain. Now Define a dictionary that has names of people in your family or friend group, including yourself, with their age, hairstyle, and pet type. (Use 'none' if they don't have a pet)

A dictionary is defined similarly to other variables, with the '"variable name" = '. Instead of using regular brackets, you use {}. Key words and their corresponding data are defined as such: '"keyword":"data"'. You access the elements by printing the dictionary, or for specific elements you write 'print("variable name"["keyword"] )'

**E.4** What is a dictionary "key" in python? Can key names be changed once they are created? Demonstrate. Say you spelled your family member's name wrong, how can you correct it? What about their pet type?

A dictionary key is the callable object that functions similarly to indexing. In my dictionary above, the keys are "Mom", "Piper", "Marissa", and "Me". Key names are not supposed to be changed once created, but you could potentially do so by completely redefining the whole data set.

```python
FriendsandFam["Momma"] = FriendsandFam.pop("Mom")
FriendsandFam
```

It's more simple to change data within the keys. You just rewrite the key with the correct data, and that overwrites the old version.

```python
FriendsandFam["Momma"]["Pets"] = ("2 cats and 1 dog")
print(FriendsandFam["Momma"])
```
