**Comparisons**

Use '==' to notate that two variables are equal to each other
Use '>' or '<' to notate greater than or less than 
- you can also do greater/less than or equal to with '>=' or '<='. The equals sign should always be on the right
Typically, you want to do this with two variables of the same type (i.e. you can't test if a str object is equal to an int object. This holds true for all comparisons)
- you can compare a float and an int, or an int and a bool, since these all have numeric notations associated with them.
You can also do the same with arrays.

**Boolean Operators**

Use 'and' to produce a "True" result only if all comparisons in the 'and' statement are true.
Use 'or' to produce a "True" result if at least one of the comparisons in the 'or' statement is true
Use 'not' to flip the boolean statement (True to False or False to True)
For arrays, you have to use the corresponding array commands:
- and -> np.logical_and()
- or -> np.logical_or()
- not -> np.logical_not()
The comparison in these arrays are seperated by commas instead of the actual "and", "or", or "not" statements

**If, Else, and Elif**

If:
- if a certain thing is true, then perform this action
Else:
- if the thing in the "if" statement is not true, perform this section instead
Elif
- if the thing in the "if" statement is not true, give another "if" statement that can potentially be met, and another action to perform instead

**Filtering DataFrames**

3 steps 
1. select column for analysis (data["column"])
2. do the comparison on the column (data["column"] > #)
3. subset df (data[data["column"] > #)
