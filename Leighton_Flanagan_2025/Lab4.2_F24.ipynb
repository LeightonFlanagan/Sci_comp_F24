**While Loops**

Basically like "if" statements that loop through data. 
Example:

 offset = 8
 while offset != 0:
     offset - offset -1
     print(offset)

This tells the program to take a variable starting at 8, subtract it by one, and print the result for every possible value of that variable until it reaches 0, at which the loop will end.
You can also add if statements in the while loop to further add conditions to the loop.

**For Loops**

For each number in a list, apply the same code within the for loop.
Example:

areas = [9, 10, 11, 12]
for i in areas:
    print(i)

This tells the program to start at index 0 in the list (i = 0), and print that, then go and do the same for each i one by one until the max i is reached. 
You can also add the indices as part of the print statement using a code like this:

areas = [9, 10, 11 ,12]
for i, a in enumerate(areas):
    print("room" + str(i + 1) + ":" + str(a)

This will give the indices in the room (displaced by one, because in typical speech "room 0" doesn't really make sense) as well as their areas. Here's an example of the output:

"room 1: 9
 room 2: 10 ..."

 And so forth

 Looping over data structures is slightly different.

**Looping over a Dictionary**

To loop over a dictionary, you need to use the items() method. 
Example:

europe = {'spain':'madrid', 'france':'paris', 'germany':'berlin',
          'norway':'oslo', 'italy':'rome', 'poland':'warsaw', 'austria':'vienna' }
for k,v in europe.items():
    print("the capital of " + k + " is " + v)

**Looping over a NumPy Array**

1D arrays use the simple version of the loop. However, with a 2D array, you need to use the np.nditer("your_array") method
Example:

for i in np.nditer(np_baseball):
    print(i)

Interestingly, this prints out the entire height column and then the entire weight column, so it doesn't retain which variables go together in this way.

**Looping over a DataFrame**

This is typically done with the iterrows() method. 
Example:

for l,r in cars.iterrows():
    print(l)
    print(r)

This prints out the label of the row (in this case, the country abbreviation) as well as the contents of each row. You can also select variables from the DataFrame that you want included.
Example:

for lab, row in cars.iterrows() :
    print(lab + ": " + str(row['cars_per_cap']))

This only prints the label and the value in the 'cars_per_cap' column. You can also add columns using the for loop. 
Example: 

for lab, row in cars.iterrows() :
    cars["COUNTRY"] = cars["country"].apply(str.upper)
print(cars)

This adds a column 'COUNTRY' that has all of the country names in uppercase. You can also use .iterrows but this is less efficient.
