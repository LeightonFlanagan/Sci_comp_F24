**Random Walks**

This basically doing a chain of these random number scenarios using a for loop. 
Example: 

random_walk = [0]
for x in range(100):
 step = random_walk[-1]
 dice = np.random.randint(1,7)
 if dice <= 2:
        step = step - 1
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1,7)
 random_walk.append(step)

This code finds a random number 1 through 6 that you would roll on a die, then assigns certain actions to each number that can possibly be rolled. This could all be done with the random numbers code above. What makes it a walk is this line: for x in range(100):, which tells python to iterate over the random number code 100 times. The random_walk.append(step) adds each new roll to random_walk so that all 100 rolls are saved in the list to view the progression from roll 1 to roll 100.

This code, however, allows the step counter to go below zero, which in real life you can't do. To make sure these loops don't allow going below a certain number, you can use the max() function. The code will be updated like this:

 **Distributions**

In order to create distributions in pythons, you do random walks multiple times. Basically, you add another for loop over the one for your random walk to iterate through the walk however many times you would like. You can then plot a histogram of your final result. Here's an example code: 

all_walks = []
for i in range(500) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.001 :
            step = 0
        random_walk.append(step)
    all_walks.append(random_walk)

np_aw_t = np.transpose(np.array(all_walks))

ends = np_aw_t[-1,:]

plt.hist(ends)
plt.show()

This loops over the random walk 500 times, assimilating all of those lists in to the variable all_walks. To plot the data, all_walks needs to be turned into an array and transposed for readability. The last row only should be plotted because this is the final step that the person in the simulation made it to. I will also give a full example of this code in a cell below the cheat sheet. 


```python
np.random.seed(123)
random_walk = [0]
for x in range(100):
 step = random_walk[-1]
 dice = np.random.randint(1,7)
 if dice <= 2:
     step = max(0, step - 1)
 elif dice <= 5:
     step = step + 1
 else:
     step = step + np.random.randint(1,7)
 random_walk.append(step)
print(random_walk)

plt.plot(random_walk)
plt.show()
```

```python
all_walks = []
for i in range(500) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.001 :
            step = 0
        random_walk.append(step)
    all_walks.append(random_walk)

np_aw_t = np.transpose(np.array(all_walks))

ends = np_aw_t[-1,:]

plt.hist(ends)
plt.show()
```
