**Intro to Functions**

To write your own function in python you need to define it by giving it a call name and what the function does. You can also specify a docstring, which is written within 3 sets of double quotes, to explain the use of the function when called upon with a '?'. 



```python
#Simple function with docstring
def shout():
    """Print a string with three exclamation marks"""
    shout_word = ('congratulations' + '!!!')
    print(shout_word)

shout()
```

You can also create a function to take any input by modifying the name to contain a parameter within the parentheses: shout(word)

```python
#Parameter function
def shout(word):
    shout_word = word + '!!!'
    print(shout_word)

shout('congratulations')
```

Lastly, instead of printing within the function, you can return a value that can be printed manually.

```python
#Returning a value
def shout(word):
    shout_word = word + '!!!'
    return(shout_word)

yell = shout('congratulations')
print(yell)
```

**Multiple Parameters and Tuples**

Functions are able to have multiple parameters by adding an extra parameter when defining the function, and then tacking on a code for both parameters. 

```python
def shout(word1, word2):
    """Concatenate strings with 3 exclamations marks"""
    shout1 = word1 + '!!!'
    shout2 = word2 + '!!!'
    new_shout = shout1 + shout2
    return new_shout

yell = shout('congratulations','you')
print(yell)
```

Tuples are like a list, but instead defined as parentheses. Tuples can be unpacked into seperate variables. 

```python
tuple1 = (2,4,6)
num1, num2, num3 = tuple1
print(num1)
print(num2)
print(num3)
```

We can use tuples in the multiple parameter code from before, and then print out the two words as seperate values.

```python
def shout_all(word1, word2):
    """Return a tuple of strings"""
    shout1 = word1 + '!!!'
    shout2 = word2 + '!!!'
    new_shout = shout1, shout2
    return new_shout

yell1, yell2 = shout_all('congratulations','you')
print(yell1)
print(yell2)
```
