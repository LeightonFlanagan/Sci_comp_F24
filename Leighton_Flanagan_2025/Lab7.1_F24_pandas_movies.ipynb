**W.1** Make a map of Antarctica using a polar projection in cartopy. Add a marker for McMurdo Station.

```python
plt.figure(figsize=(10, 10))
ax = plt.axes(projection=ccrs.SouthPolarStereo())
ax.set_extent([-180,180,-90,-60], crs=ccrs.PlateCarree())
ax.coastlines(resolution='110m')
ax.gridlines()
ax.annotate("McMurdo Station", [-135, -68], transform=ccrs.PlateCarree(), color = 'red', size = 15)
ax.plot([-166.3952], [-77.5111], marker = '*', transform=ccrs.PlateCarree(), linestyle = 'None', markersize = 20, markerfacecolor = 'red', markeredgecolor = 'black')
```

### Chapter 1

**Functions**

A couple functions you can use to learn more about a pandas DataFrame:

.head(): returns first few rows (head) of DataFrame
.info(): shows info about each of the columns, like the data type
.shape: returns the numbers of rows and columns
.describe(): calculates summary statistics for each column
.values: prints just the values of the array (no indices)
.columns: returns column names
.index: returns the row index (either numbers or names)

**Sorting and Subsetting DataFrames**

To sort rows by a particular column, use df.sort_values("column_name")

For descending sorting, use df.sort_values("column_name", ascending = False)

For sorting by multiple columns, use df.sort_values(["column1", "column2"], ascending = [True, False])

To subset columns, just call up df["column_name"]

To subset multiple columns, call up df[["column1", "column2"]]

To subset rows, call up df[df["column"] > or < or == #]]

To subset multiple rows, call up df[(df["column1"] > or < or == #), (df["column2"] > or < or == #)]

To subset rows categorically, specify a variable with the values you're looking for, then use df[df["column"].isin(variable)]

**Adding Columns**

To add a column, just write df[new_column_name] = whatever you want to define the column as 

### Chapter 2

**Summary Statistics**

To get the mean of a DataFrame column, use df["column"].mean()

To get the median of a DataFrame column, use df["column"].median()

To get the max of a DataFrame column, use df["column"].max()

To get the min of a DataFrame column, use df["column"].min()

To use a function you create yourself on a DataFrame column, write df["column"].agg(function)

To get cumulative statistics, use .cumsum(), .cummax(), .cummin(), and .cumprod()

**Counting**

To drop duplicates, use df.drop_duplicates(subset = columns_you_want_this_applied_to)

To count, use df[column].value_counts(sort = (to sort), normalize = (to get proportions))

**Grouped Summary Statistics**

For more concise coding when getting statistics within a DataFrame column, use df.groupby(groups)[stats_column].stat_function

**Pivot Tables**

Another method for grouped summary statistics

df.pivot_table(values = "column1", index = "column2", columns = "column 3", fill_value = 0, margins = True(adds sums of all rows and columns to the outside of table))
