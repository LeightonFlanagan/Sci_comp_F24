### Chapter 3

**Explicit Indexes**

To set a particular column as an index for your DataFrame, use 

    df.set_index("column")
To reset the index, use

    df.reset_index()
    df.reset_index(drop = True <- drops the column from the df)
To subset within your index, use .loc() for simpler code:

    df.loc[["term1", "term2]]
For multi-level indexes, use

    df.set_index(["index1", "index2"])
You can also create a list of things to subset within your indexes

    rows_to_keep = [("term1_from_index1", "term1_from_index2"), ("term2_from_index1",  "term2_from_index2")]
    df.loc[rows_to_keep]
To sort an index, use

    df.sort_index()
    df.sort_index(level = "index_of_multi_index_you_want_to_sort")
Ascending/descending still works the same way as .sort_values()

**Slicing Indexes**

Note that indexes can only be sliced when they are sorted. To slice the primary index of a DataFrame, use

    df.loc["term1":"term2"]
To slice the secondary index of a DataFrame, you also need to include the primary index.

    df.loc[("term1_from_index1", "term1_from_index2"):("term2_from_index1", "term2_from_index2")]
You can also slice columns at the same time as indexes

    df.loc[("term1_from_index1", "term1_from_index2"):("term2_from_index1", "term2_from_index2"), "column1":"column2"]
To slice within a time series, use

    df.loc["year1":"year2"]
    df.loc["yyyy-mm-dd1":"yyyy-mm-dd2"]
To subest by index number, use .iloc instead of .loc

    df.iloc[#:#, #:#]

**Pivot Tables**

Side note: to get the year/month/day out of a date_time object, use

    df["column"]dt.year/month/day
Subsetting pivot tables is the same as a regular DataFrame. Just make sure you know what your new indexes and columns are.

To do calculations within a pivot table

    df.mean()
    df.mean(axis = "columns")
Note that you use the opposite of what you're trying to find the mean value for. For example, to find the mean value by the columns, use .mean(). To find the mean value for the index, use .mean(axis = "columns")

To filter those means, use

    df[df == df.max()]
    df[df == df.min()]

**Visualiziing Data**

To create a bar plot of the sum of values grouped by a column, use

    df_grouped = df.groupby("category_column")["data_column"].sum()
    df_grouped.plot(kind = "bar")
This creates as many bars as there are categories in the grouping column, and plots the sum of the values in the data column for the height of the bars

You can do the same thing with a line plot:

    df_grouped = df.groupby("date/time_column")["data_column"].sum()
    df_grouped.plot(kind = "line")
To create a scatter plot to show the relationship between two groups, use

    df.plot(x = "group1", y = "group2", kind = "scatter", title = "title")
To create a histogram comparing two different categories, use

    df[df["category_column"] == "category1"]["values_column"].hist()
    df[df["category_column"] == "category2"]["values_column"].hist()
    plt.legend(["category1","category2"])
Here, you can specify the alpha value and the number of bins

    df[df["category_column"] == "category1"]["values_column"].hist(alpha = 0.5, bins = 20)

**Missing Values**

To find missing values, use

    df.isna() <- for each individual value
    df.isna().any() <- organized by columns
To remove rows with missing values, use

    df.dropna()
To replace missing values, use 

    df.fillna(0)

**Creating DataFrames**

To make a DataFrame, you can either create a list of dictionaries or a dictionary of lists

List of dictionaries:

    list1 = [{"key1":value1, "key2":value2}, {"key1.1":value1.1, "key2.1":value2.1}]
Dictionary of lists

     list2 = {
     "key1":[value1, value2]
     "key2":[value1.1, value2.1]
     }
To turn these into a DataFrame:

    pd.DataFrame(list1)
