**W.1** Write a function that takes a list and returns a new list that contains all the elements of the first list minus all the duplicates.

```python
def list_no_duplicates(inlist):
    outlist = []
    for i in range(1, max(inlist)+1):
        if i in inlist: 
            if i not in outlist:
                outlist.append(i)
    return outlist
```

```python
test = list_no_duplicates([1, 1, 3, 4, 4, 5, 16, 1, 7, 4, 3])
```
